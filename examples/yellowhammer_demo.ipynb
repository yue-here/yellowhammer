# _datalab_ yellowhammer agent demo

This demo uses the yellowhammer AI agent query a *datalab* instance populated by data from Josh Bocarsly's group at the University of Houston.


The agent requires access to a LLM provider. We can first specify the model we want to use, and the API key. This can be done by setting environment variables or through a `.env` file, but here we do it explicitly in the notebook.

In this demo, we use the Google Gemini model which has a generous free tier. You can get a free API key here: https://aistudio.google.com/app/apikey

Any model supported by Pydantic AI can be used. The list of model names is here: https://ai.pydantic.dev/api/models/base/

```python
import os
import getpass

# Specify the AI model
os.environ["AI_MODEL"] = "google-gla:gemini-2.5-flash"

# Use the correct environment variable name for the API key
# os.environ["GEMINI_API_KEY"] = getpass.getpass("Enter your API key:")
```

We now need to connect to a _datalab_ instance using its API key. Here we connect to a locally hosted instance, but normally you will connect to a remote instance such as `https://demo-api.datalab-org.io`.

```python
# Set the Datalab API key and URL
# os.environ["DATALAB_API_KEY"] = getpass.getpass("Enter your Datalab API key")
# os.environ["DATALAB_URL"] = "http://api.localhost:5001/"
os.environ["DATALAB_URL"] = "http://localhost:5001/"

# Load the yellowhammer extension
%reload_ext yellowhammer
```

Now we can start using the `%%llm` cell magic to talk to _datalab_!

```python
%%llm
Can you given an overview of Moinak Dutta's research?
```

```python
%%llm
tell me about IMG_1648.jpg
```
